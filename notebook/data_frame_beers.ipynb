```python
#import necessary modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# convert the list of beers links to a DataFrame
df = pd.read_excel('../beers_spain.xlsx')
```

```python
#print the DataFrame
df
```

```python
# remove rows with NaN values in specific columns
df = df.dropna(subset=['Name', 'Main Price', 'lot Size', 'Supermarkets and Prices'])
df = df.reset_index(drop=True)
df
```

```python
#edit the 'lot Size' column to remove the '/' character
df['lot Size'] = df['lot Size'].str.replace('/', '',)
df
```

```python
# remove columns that are not needed
df = df.drop(['Product Link', 'Image'], axis=1)
df
```

```python
# modify the 'Main Price' column to remove the Euro sign and strip whitespace and convert to numeric
df['Main Price'] = df['Main Price'].str.replace('€', '').str.replace(',','.').str.strip()
df['Main Price'] = pd.to_numeric(df['Main Price'], errors='coerce')
df
```

```python
# remove rows with 'sidra', 'vino', 'cider', 'colágeno' and 'vermut' in the 'Name' column
clean = (
    (df['Name'].str.contains('sidra', case=False, na=False) == False) &
    (df['Name'].str.contains('vino', case=False, na=False) == False) &
    (df['Name'].str.contains('cider', case=False, na=False) == False) &
    (df['Name'].str.contains('colágeno', case=False, na=False) == False) &
    (df['Name'].str.contains('vermut', case=False, na=False) == False)
)
df = df.loc[clean].reset_index(drop=True)
df
```

```python
# replace the 'Supermarkets and Prices' column to remove the Euro sign and strip whitespace
df['Supermarkets and Prices'] = df['Supermarkets and Prices'].str.replace('€', '').str.replace(',','.').str.strip()
df
```

```python
# replace '\n' with '-' in the 'Name' column
df['Name'] = df['Name'].str.replace('\n', '-')
df
```

```python
# count the number of beers that have 'botella' or 'botellín' in the 'lot Size' column
count_1 = df['lot Size'].str.contains('botella', case=False, na=False).sum() + \
          df['lot Size'].str.contains('botellín', case=False, na=False).sum()
print (f"Number of beers in 'Botella' or 'Botellín': {count_1}")

# count the number of beers that have 'lata' in the 'lot Size' column
count_2 = df['lot Size'].str.contains('lata', case=False, na=False).sum()
print (f"Number of beers in 'Lata': {count_2}")

# print the hypothesis based on the counts
if count_1 > count_2:
    print("Correct hypothesis: in Spanish supermarkets, the most common format for selling beer is 'Botella' or 'Botellín'.")
elif count_2 > count_1:
    print("Incorrect hypothesis: in Spanish supermarkets, the most common format for selling beer is 'Lata'.")
else:
    print("Both formats have the same frequency.")
```

```python
# Export the cleaned DataFrame to Excel
df.to_excel('beers_spain_modify.xlsx', index=False)
```

```python
#  Plotting the results. Testing the hypothesis
sns.barplot(x=['Botella/Botellín', 'Lata'], y=[count_1, count_2], palette=['#145A32', '#FFD700'])
plt.ylabel('Cantidad')
plt.title('Formato más común para la venta de cervezas') 
plt.show()
```

```python
#Barplot of the 10 most frequent lot size formats
top10 = df['lot Size'].value_counts().nlargest(10)

plt.figure()
sns.barplot(
    x=top10.values,
    y=top10.index,
    palette='mako'
)
plt.title('Los 10 formatos de tamaño de lote más comunes')
plt.xlabel('Frecuencia')
plt.ylabel('Formato de tamaño de lote')
plt.tight_layout()
plt.show()
```

```python
#Pie chart of the top 5 most common lot size formats
top5 = df['lot Size'].value_counts().nlargest(5)
plt.figure(figsize=(6,6))
plt.pie(
    top5.values,
    labels=top5.index,
    autopct='%1.1f%%',
)
plt.title('Los 5 formatos de tamaño de lote (porcentaje)')
plt.tight_layout()
plt.show()
```

```python
# Main Price comparing latas vs. botellas
plt.figure(figsize=(8,5))
sns.kdeplot(
    data=df[df['lot Size'].str.contains('lata', case=False, na=False)],
    x='Main Price',
    label='Lata',
    fill=True,
    alpha=0.4
)
sns.kdeplot(
    data=df[df['lot Size'].str.contains('botell', case=False, na=False)],
    x='Main Price',
    label='Botella/Botellín',
    fill=True,
    alpha=0.4
)
plt.title('Distribución de precios: Latas vs. Botellas/Botellines')
plt.xlabel('Precios (€)')
plt.ylabel('Densidad')
plt.legend()
plt.tight_layout()
plt.show()
```
